```python
#Seasonal analysis
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Autumn"

df4["Season"] = df4["Month"].apply(get_season)
```

```python
df4.groupby("Season")["Temperature"].agg(["mean", "min", "max"])
```

```python
#Hottest, coldest and wettest
hottest_day = df4.loc[df4["Temperature"].idxmax()]
coldest_day = df4.loc[df4["Temperature"].idxmin()]
wettest_day = df4.loc[df4["Rainfall"].idxmax()]

print("Hottest Day:")
print(hottest_day)

print("\nColdest Day:")
print(coldest_day)

print("\nWettest Day:")
print(wettest_day)
```
