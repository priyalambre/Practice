```python
# cities v/s Average tour price in inr
# ============================================================
city_price = df.groupby("City")["Tour_Price_INR"].mean().sort_values()

plt.figure(figsize=(30,12))
plt.plot(city_price.index, city_price.values)

plt.bar(city_price.index, city_price.values)

plt.xlabel("Average Tour Price (INR)")
plt.ylabel("City")
plt.title("Average Tour Price by City")

plt.show()
```

```python
# Country v/s tour price
# ============================================================
country_price = df.groupby('Country')['Tour_Price_INR'].mean()

plt.figure(figsize =(20,12))

plt.plot(country_price.index, country_price.values)
plt.bar(country_price.index, country_price.values)

plt.show()
```

```python
# TOUR TYPE VS AVERAGE TOUR PRICE
# ============================================================

tour_type_price = df.groupby("Tour_Type")["Tour_Price_INR"].mean()

plt.figure(figsize=(8, 5))

plt.bar(tour_type_price.index, tour_type_price.values)

plt.xlabel("Tour Type")
plt.ylabel("Average Tour Price (INR)")
plt.title("Average Tour Price by Tour Type")

plt.show()
```

```python
#SEASON VS AVERAGE TOUR PRICE
# ============================================================

season_price = df.groupby("Season")["Tour_Price_INR"].mean()

plt.figure(figsize=(8, 5))

plt.bar(season_price.index, season_price.values)

plt.xlabel("Season")
plt.ylabel("Average Tour Price (INR)")
plt.title("Average Tour Price by Season")

plt.show()
```

```python
# TOUR DURATION VS AVERAGE TOUR PRICE
# ============================================================

duration_price = df.groupby("Duration_Days")["Tour_Price_INR"].mean()

plt.figure(figsize=(10, 6))

plt.plot(
    duration_price.index,
    duration_price.values,
    marker="o"
)

plt.xlabel("Duration (Days)")
plt.ylabel("Average Tour Price (INR)")
plt.title("Tour Duration vs Average Tour Price")

plt.grid(True)

plt.show()
```

```python
# CITY VS TOTAL BOOKINGS
# ============================================================

city_bookings = df.groupby("City")["Bookings"].sum()

city_bookings = city_bookings.sort_values(ascending=False)

plt.figure(figsize=(16, 12))

plt.barh(city_bookings.index, city_bookings.values)

plt.xlabel("Total Bookings")
plt.ylabel("City")
plt.title("Total Bookings by City")

plt.gca().invert_yaxis()

plt.show()
```

```python
# TOUR TYPE VS AVERAGE BOOKINGS
# ============================================================

tour_type_bookings = df.groupby("Tour_Type")["Bookings"].mean()

plt.figure(figsize=(8, 5))

plt.bar(
    tour_type_bookings.index,
    tour_type_bookings.values
)

plt.xlabel("Tour Type")
plt.ylabel("Average Bookings")
plt.title("Average Bookings by Tour Type")

plt.show()
```

```python
# SEASON VS TOTAL BOOKINGS
# ============================================================

season_bookings = df.groupby("Season")["Bookings"].sum()

plt.figure(figsize=(8, 5))

plt.bar(
    season_bookings.index,
    season_bookings.values
)

plt.xlabel("Season")
plt.ylabel("Total Bookings")
plt.title("Total Bookings by Season")

plt.show()
```

```python
# COST COMPONENT ANALYSIS
# ============================================================

cost_columns = [
    "Hotel_Cost",
    "Flight_Cost",
    "Food_Cost",
    "Activities_Cost",
    "Local_Transport_Cost"
]

average_costs = df[cost_columns].mean()

plt.figure(figsize=(10, 6))

plt.bar(
    average_costs.index,
    average_costs.values
)

plt.xlabel("Cost Component")
plt.ylabel("Average Cost (INR)")
plt.title("Average Tour Cost Components")

plt.xticks(rotation=60)

plt.show()
```
