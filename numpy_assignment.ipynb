```python
#Exercises

# Columns: Age, Math, Science
students = np.array([
    [18, 85, 78],
    [19, 92, 88],
    [20, 76, 95],
    [18, 89, 72],
    [21, 65, 80]
])
print(students)
```

```python
#Get the shape of the matrix.
print(students.shape)
```

```python
#Find the average age of students.
print("Mean students Age: ",students[:,0].mean())
```

```python
#Extract the Math marks of all students.
print('Maths Marks: ', students[:,1])
```

```python
#Find the highest Science mark.
print("Highest Marks in Science : ", students[:,2].max())
```

```python
#Get the details of the student who scored more than 90 in Math.
print("Details of students who scored more than 90 in Maths")
print(students[students[:,1]>90])
```

```python
#Increase the Math marks of all students by 5.
students[:,1] = students[:,1]+5
print(students)
```

```python
#Find how many students are younger than 19.
print(np.sum(students[:, 0] < 19))
```

```python
#Calculate the average marks in each subject (column-wise mean).
print(np.mean(students[:, 1:], axis=0))
```

```python
#Get the data of students who scored at least 80 in both subjects.
result = students[(students[:, 1] >= 80) & (students[:, 2] >= 80)]

print(result)
```

```python
#Replace all Science marks below 75 with 0.
students[students[:, 2] < 75, 2] = 0

print(students)
```
